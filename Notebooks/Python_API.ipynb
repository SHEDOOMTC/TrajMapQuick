### To demonstrate ***shift*** and ***hotspot*** subcommands

We will use a trajectory from apo strucutre of ***P. Falciparum*** cGMP-dependent protein kinase (PDB ID: 8EM8), with **790** residues 

**1. Loading all the libraries from the core module. There are ten core modules in the tmq library**

```python
#To allow the plots to show up
%matplotlib inline
```

```python
from tmq.core.traj2pdb import traj2pdb
from tmq.core.pdb2csv import pdb2csv  
from tmq.core.trajmap import trajmap
from tmq.core.csv2shift import shift
from tmq.core.shift2graph import shift2graph
from tmq.core.hotspot import detect_shifts
from tmq.core.hotspot import plot_hotspot_regions
from tmq.core.hotspot import slice_shifts
from tmq.core.hotspot import plot_slice_regions
from tmq.core.diff2graph import diff_map
from tmq.core.average_trajmap import avg_trajmap
```

**2. Loading a trajectory and converting to PDB (The trajectories used here are cut to 125 frames; to see the full size trajectories, donload from zenodo doi....)**

```python
#Loads the trajectory (125 frames) at a stride of 1
TRAJ = traj2pdb("../Test/topol_1.prmtop", "../Test/traj_1_125.nc", 1)
TRAJ.save_pdb("traj.pdb")
```

**3. Convert the PDB to a CSV of distance matrices**

```python
#compute per residue distances using the backbone atoms between residues 1 and 789
matrix = pdb2csv("traj.pdb", 1, 789, "backbone")
```

**4. Plot a trajectory map of the distance matrix**

```python
#Trajectory map for residues 1-789 and max colorbar of 5
trajmap(matrix, 1, 789, 5)
```

**5. Compute per residue shift as a measure of fluctuation for a given region of the protein (resid 18 and 30)**

```python
#computes residue shift for the 100 frames of the trajectory loaded above
res_shift = shift(matrix, 170, 200)
```

**6. Plot the shift graph for this region of the protein (resid 18 and 30)**

```python
#make a plot of the shift for the residue region across the frames (use semicolon to suppress dual figure display)
shift2graph(res_shift, "170-200")
```

**7. Auto detect regions of the proteins with the most shift**

```python
#Detect 10 hotspots in your protein using an average window of 3 and interpeak distance of 5
mean_shift, smooth, regions = detect_shifts(matrix, 3, 5, 10)
```

**8. Plot an RMSF plot to visualize the hotspot regions identified**

```python
#Visualize the hotspots regions identified
plot_hotspot_regions(mean_shift, smooth, regions, title="Residue Shift Hotspots")
```

**9. Subplots for each of the hotspot region identified**

```python
#Make plots for each of the hotspot region identified
results = slice_shifts(matrix, 3, 5, 10)
```

```python
for start, end, shift_data in results:
    fig = plot_slice_regions(shift_data, start, end)
    display (fig)
```

### To demonstrate ***diff*** and ***average*** subcommand** 

We will use apo and ligand bound ***BACE-1*** protein (PDB ID: 1XS7) with **389** residues. The trajectories used here are cut to **250 frames**; to see the full size trajectories, donload from zenodo doi....)

**1. Loading the trajectories and convert to PDB**

```python
#Loads the trajectory (250 frames) at a stride of 1 for the apo protein
TRAJ1 = traj2pdb("../Test/topol_2_apo.prmtop", "../Test/traj_2_apo.nc", 1)
TRAJ1.save_pdb("traj_apo.pdb")
```

```python
#Loads the trajectory (250 frames) at a stride of 1 for the bound protein
TRAJ2 = traj2pdb("../Test/topol_2_bound.prmtop", "../Test/traj_2_bound.nc", 1)
TRAJ2.save_pdb("traj_bound.pdb")
```

**2. Convert the PDB to a CSV of distance matrices**

```python
#compute per residue distances using the backbone atoms between residues 1 and 389 for the apo
matrix1 = pdb2csv("traj_apo.pdb", 1, 389, "backbone")
```

```python
#compute per residue distances using the backbone atoms between residues 1 and 389 for the apo
matrix2 = pdb2csv("traj_bound.pdb", 1, 389, "backbone")
```

**3. Compute the difference matrix and make a difference map plot**

```python
#pass the matrices, then the vmax and residue range
diff_map(matrix1, matrix2, 6, 1, 389)
```

**4. compute the average of two simulation and plot the average trajectory map**

```python
#pass the matrices, then the vmax and residue range
avg_trajmap(1, 389, 6, matrix1, matrix2)
```
